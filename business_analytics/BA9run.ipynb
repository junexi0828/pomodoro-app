```python
import tensorflow as tf

# tf.constant() 함수는 상수 텐서를 생성합니다.
# 여기서는 값이 1인 상수 텐서를 만들고, 데이터 타입을 float32(32비트 실수)로 지정합니다.
# dtype=tf.float32를 지정하지 않으면 기본적으로 int32로 생성됩니다.
a = tf.constant(1, dtype=tf.float32)


print(a)
```

```python
import tensorflow as tf

# 2행 2열(2x2) 실수형 행렬(텐서) 생성
# 각 원소는 float32 타입이며, shape는 (2, 2)임
a = tf.constant([[3.14, 1.7], [5.6, 2.2]], dtype=tf.float32)
print(a)  # shape=(2, 2)

# 3행 2열(3x2) 실수형 행렬(텐서) 생성
# 행의 개수(3), 열의 개수(2)로 shape가 (3, 2)임
a = tf.constant([[3.14, 1.7], [5.6, 2.2], [3, 3]], dtype=tf.float32)
print(a)  # shape=(3, 2)

# 두 행렬의 덧셈과 행렬 곱 연산 예시
# a, b 모두 2x2 실수형 행렬로 생성
a = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
b = tf.constant([[2, 2], [3, 5]], dtype=tf.float32)

# 같은 shape의 행렬끼리 원소별 덧셈(Broadcasting 아님, 각 위치의 값끼리 더함)
print(a + b)

# tf.matmul은 행렬 곱 연산을 수행
# (2,2) @ (2,2) → (2,2) 결과, 행렬 곱의 조건(앞 행렬의 열 수 = 뒤 행렬의 행 수)이 맞아야 함
print(tf.matmul(a, b))
```

```python
import tensorflow as tf

# 2x2 행렬 a 생성
a = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)

# 2x2 단위행렬(Identity matrix) 생성: 대각선이 1, 나머지는 0인 정방행렬
i = tf.constant(tf.eye(2), dtype=tf.float32)

# 행렬 a와 단위행렬 i의 곱: 결과는 항상 원래 행렬 a와 동일
print(tf.matmul(a, i))
print(tf.matmul(i, a))
```

```python
import tensorflow as tf

# 2x2 정방행렬 a 생성
a = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)

# tf.linalg.inv()를 사용해 a의 역행렬을 계산
# 역행렬은 정방행렬(행과 열의 수가 같은 행렬)에서만 존재하며,
# 원래 행렬과 곱하면 단위행렬(Identity matrix)이 됨
aInv = tf.linalg.inv(a)

# a의 역행렬 출력
print(aInv)

# a와 aInv를 곱하면 단위행렬이 출력됨
# 단, 컴퓨터의 소수점 연산 한계로 인해 완벽한 0, 1이 아닌 근사값이 나올 수 있음
print(tf.matmul(a, aInv))
```

```python
import pandas as pd
sd = pd.read_csv('softdrink.csv')

import tensorflow as tf

# 입력 변수(기온)와 타깃 변수(매출액)를 float32 타입의 텐서로 변환
x = tf.constant(sd['temp'], dtype=tf.float32)
y = tf.constant(sd['sales'], dtype=tf.float32)

print(x)
print(y)

# 선형회귀 가중치(a)와 절편(b)를 변수로 선언 (학습 전 임의 초기값)
a = tf.Variable(20, dtype=tf.float32)
b = tf.Variable(650, dtype=tf.float32)

# 예측값 계산: y_hat = a * x + b (선형회귀의 가설 함수)
y_hat = a * x + b

print(y_hat)

import matplotlib.pyplot as plt

# 실제 데이터 산점도(파란색)와 예측 직선(빨간색) 시각화
plt.scatter(x, y, color='blue')   # 실제 데이터
plt.plot(x, y_hat, color='red')   # 예측 직선
plt.show()
```

```python
import pandas as pd
import tensorflow as tf

# 선형회귀 예측 함수 정의: 입력 x, 가중치 a, 절편 b를 받아 예측값 반환
def y_hat(x, a, b):
    return a * x + b

# 데이터 읽기 및 텐서 변환
sd = pd.read_csv('softdrink.csv')
x = tf.constant(sd['temp'], dtype=tf.float32)    # 입력 변수(기온)
y = tf.constant(sd['sales'], dtype=tf.float32)   # 타깃 변수(매출액)

# 가중치(a)와 절편(b)를 변수로 선언 (초기값 임의 지정)
a = tf.Variable(20, dtype=tf.float32)
b = tf.Variable(650, dtype=tf.float32)

# 예측값 계산 및 출력
print(y_hat(x, a, b))
```

```python
# 평균제곱오차(MSE) 계산 함수 정의
def loss(a, b, x, y):
    # (예측값 - 실제값)의 제곱의 평균을 반환 (MSE)
    return tf.reduce_mean(tf.square(y - y_hat(x, a, b)))
    # tf.reduce_mean: 텐서의 원소별 평균값을 계산하는 텐서플로우 함수

# --- 아래는 예시 코드 (ex9-2.py) ---

import pandas as pd
import tensorflow as tf

# 데이터 읽기 및 텐서 변환
sd = pd.read_csv('softdrink.csv')
x = tf.constant(sd['temp'], dtype=tf.float32)
y = tf.constant(sd['sales'], dtype=tf.float32)

# 가중치(a)와 절편(b) 초기값 설정
a = tf.Variable(20, dtype=tf.float32)
b = tf.Variable(650, dtype=tf.float32)

# MSE(평균제곱오차) 계산 및 출력
print('MSE = ', loss(a, b, x, y))
```

```python
# tf.GradientTape를 사용해 파라미터 a, b에 대한 손실함수의 그래디언트(미분값)를 계산
with tf.GradientTape() as t:
    # 테이프에 미분하려는 연산(손실함수 계산)을 기록
    current_loss = loss(a, b, x, y)

# a, b 각각에 대한 손실함수의 미분값(그래디언트) 계산
da, db = t.gradient(current_loss, [a, b])

print('da = ', da)  # a에 대한 그래디언트
print('db = ', db)  # b에 대한 그래디언트
```

```python
while True:
    # GradientTape로 손실함수의 그래디언트(미분값) 계산
    with tf.GradientTape() as t:
        current_loss = loss(a, b, x, y)
        da, db = t.gradient(current_loss, [a, b])
    # 현재 파라미터, 손실, 그래디언트 값 출력(학습 과정 모니터링)
    print('MSE, a, b, da, db = ',
          a.numpy(), b.numpy(), loss(a, b, x, y).numpy(), da.numpy(), db.numpy())
    # 경사하강법(gradient descent)으로 파라미터 업데이트
    a.assign_add(-0.001 * da)
    b.assign_add(-0.001 * db)
    # 그래디언트가 충분히 작아지면(수렴) 반복 종료
    if abs(da) < 0.05 and abs(db) < 0.05:
        break

import matplotlib.pyplot as plt
# 실제 데이터 산점도(파란색)와 학습된 모델의 예측 직선(빨간색) 시각화
plt.scatter(x, y, color='blue')
plt.plot(x, y_hat(x, a, b), color='red')
plt.show()
```

회귀분석의 오차함수로 MSE를 사용하였음
    – MSE는 2차항이 포함된 비선형 함수(nonlinear function)
    – 이 함수를 경사하강법으로 최적화하였음

• 비선형 최적화(nonlinear optimization): 머신러닝의 근간
    – 경사하강법은 비선형 최적화 방법 중 하나
