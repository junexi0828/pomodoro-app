```python
# 리스트 한번에 전체 컬럼 할당
import pandas as pd
from ast import literal_eval
import numpy as np

# 영화 데이터 읽기 (제목, 장르만 사용)
movie_full_data = pd.read_csv('tmdb_5000_movies.csv')
movie = movie_full_data[['original_title', 'genres']]

# 장르 정보를 파싱하여 리스트로 변환
for i in range(len(movie)):
    mv = movie.iloc[i]
    mv['genres'] = literal_eval(mv['genres'])
    mv['genres'] = [y['name'] for y in mv['genres']]

# 전체 장르 목록 추출 (중복 없이)
all_genres = []
for i in range(len(movie)):
    mv = movie.iloc[i]
    for g in mv['genres']:
        if g not in all_genres:
            all_genres.append(g)
print(all_genres)

# 각 영화별로 장르 벡터(gvec) 생성 (장르가 있으면 1, 없으면 0)
gvec_list = []
for i in range(len(movie)):
    mv = movie.iloc[i]
    vec = []
    for g in all_genres:
        if g in mv['genres']:
            vec.append(1)
        else:
            vec.append(0)
    gvec_list.append(vec)
movie['gvec'] = gvec_list  # 한 번에 전체 컬럼 할당

# 벡터의 크기(노름) 계산 함수
def magnitude(v):
    m = 0
    for x in v:
        m += x**2
    return np.sqrt(m)

# 두 벡터의 코사인 유사도 계산 함수
def similarity(a, b):
    dot = np.dot(a, b)
    if dot == 0:
        return 0
    ma = magnitude(a)
    mb = magnitude(b)
    return dot / (ma * mb)

# 영화 번호를 입력받아 비슷한 영화를 추천
def recommend():
    while True:
        movie_num_str = input("\nEnter movie number: ")
        if movie_num_str == "":
            break
        movie_num = int(movie_num_str)
        print("You selected: ", movie.loc[movie_num, 'original_title'])
        my_vec = movie.loc[movie_num, 'gvec']
        sim_list = []
        for i in range(len(movie)):
            if i == movie_num:
                sim_list.append(0)
            else:
                mv = movie.iloc[i]
                sim = similarity(my_vec, mv['gvec'])
                sim_list.append(sim)
        y = np.argsort(sim_list)[::-1]
        print("Recommended:")
        for i in y[:10]:
            print('%4d'%i, '%.6f'%sim_list[i], movie.loc[i, 'original_title'])

# 추천 함수 실행
def main():
    recommend()

if __name__ == "__main__":
    main()
```
