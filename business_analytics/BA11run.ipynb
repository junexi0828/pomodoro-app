```python
# 공식 참고
sum = 0
for i in range(0, n):
    sum += w[i] * x[i]  # 각 입력(x)과 가중치(w)의 곱을 모두 더함(가중합)

# 가중합이 임계값(threshold) 이상이면 신호를 다음 뉴런으로 전달
if sum >= threshold:
    transmit_to_other_neurons()
```
